Используйте код из семинара для нахождения кандидатов в дубликаты (шинглы -> минхэш - lsh) и рассчитайте реальную меру Жаккара между полученными кандидатами. Настройте параметры k, num_hash_functions, bands так чтобы результаты получались адекватные (мера Жаккара хотя бы больше нуля).

```python
def get_shingles(text, k=5):
    """генерирует список шинглов из строки (символьные шинглы)"""
    shingles = set()
    for i in range(len(text) - k + 1):
        shingle = text[i:i + k]
        shingles.add(shingle)
    return shingles
```

```python
import hashlib
def hash_string(s):
    """хеширует строку и возвращает число"""
    return int(hashlib.md5(s.encode('utf8')).hexdigest(), 16) 
```

```python
def generate_hash_functions(k):
    """генерирует k хеш-функций добавляя индекс к строке"""
    functions = []
    for i in range(k):
        functions.append(lambda x, i=i: hash_string(x + str(i)))
    return functions
```

```python
def compute_minhash_signature(shingles, hash_funcs):
    """вычисляет minhash-сигнатуру для списка шинглов"""
    signature = []
    for hash_func in hash_funcs:
        if not shingles:
            signature.append(2**128 - 1)
            continue
        min_hash = min(hash_func(shingle) for shingle in shingles)
        signature.append(min_hash)
    return signature
```

```python
def find_similar_strings(strings_list, k=5, num_hashes=100, bands=20):
    hash_funcs = generate_hash_functions(num_hashes)
    signatures = []
    shingles_list = []

    # каждый текст обрабатывается отдельно: шинглы и сигнатуры
    for string in tqdm(strings_list, desc="MinHash (chunks)"):
        shingles = get_shingles(string, k)
        shingles_list.append(shingles)
        signature = compute_minhash_signature(shingles, hash_funcs)
        signatures.append(signature)

    # вычисляются кандидаты по кускам сигнатур
    buckets = lsh(signatures, bands)
    candidates = set()
    for bucket_items in buckets.values():
        if len(bucket_items) > 1:
            for i in bucket_items:
                for j in bucket_items:
                    if i < j:
                        candidates.add((i, j))

    return candidates, signatures, shingles_list
```

```python
def get_jaccard(x, y, k=10):
    x_sh = get_shingles(x, k)
    y_sh = get_shingles(y, k)
    if not x_sh or not y_sh:
        return 0.0
    return len(x_sh & y_sh) / len(x_sh | y_sh)
```

```python
from lxml import etree

def extract_text_from_fb2(path: str) -> str:
    parser = etree.XMLParser(recover=True)
    tree = etree.parse(path, parser)
    root = tree.getroot()
    ns = {"fb": "http://www.gribuser.ru/xml/fictionbook/2.0"}
    ps = root.xpath('//fb:body//fb:p', namespaces=ns)
    return " ".join((p.text or "").strip() for p in ps if p.text)


def split_into_sliding_windows_words(text: str, window_size=200, step=200):
  
    words = text.split()
    if not words:
        return []
    chunks = []
    for start in range(0, max(1, len(words)-window_size+1), step):
        win = words[start:start+window_size]
        chunks.append(" ".join(win))
    if not chunks:
        chunks.append(" ".join(words))
    return chunks
```

```python
import itertools

def test_lsh(raw_text, param_combinations):
    
    all_results = []
    
    for params in param_combinations:
        print(f"\n{'='*80}")
        print(f"ТЕСТ: {params}")
        print(f"{'='*80}")
        
        windows = split_into_sliding_windows_words(
            raw_text, 
            window_size=params['window_size'],
            step=params['window_size']
        )
        
        print(f"Создано окон (без перекрытия): {len(windows)}")
        if len(windows) > 0:
            print(f"Пример окна: {windows[0][:100]}...")
        
        candidates, signatures, shingles_list = find_similar_strings(
            strings_list=windows,
            k=params['k'],
            num_hashes=params['num_hashes'],
            bands=params['bands']
        )
        
        print(f"Кандидатов найдено: {len(candidates)}")
        
        flat_pairs = set()
        if isinstance(candidates, set):
            for pair in candidates:
                if isinstance(pair, (list, tuple)) and len(pair) == 2:
                    flat_pairs.add(tuple(sorted(pair)))
        
        # Вычисляем Jaccard
        similar_pairs = []
        for pair in flat_pairs:
            i, j = pair
            if abs(i - j) > 1: 
                jac = get_jaccard(windows[i], windows[j])
                if jac > 0:
                    similar_pairs.append((i, j, jac))
        
        similar_pairs.sort(key=lambda x: -x[2])
        
        # Сохраняем результаты
        result = {
            'params': params,
            'windows_count': len(windows),
            'candidates_count': len(candidates),
            'similar_pairs_count': len(similar_pairs),
            'similar_pairs': similar_pairs[:10],  # Топ-10 пар
            'max_jaccard': similar_pairs[0][2] if similar_pairs else 0
        }
        
        all_results.append(result)
        
        # Выводим результаты
        print(f"Похожих пар (J>0): {len(similar_pairs)}")
        print(f"Максимальный Jaccard: {result['max_jaccard']:.4f}")
        
        if similar_pairs:
            print("Топ-5 похожих пар:")
            for i, j, jac in similar_pairs[:5]:
                print(f"  {i} — {j} | J={jac:.4f}")
                # Показываем пример для лучшей пары
                if jac == result['max_jaccard']:
                    print(f"    Окно {i}: {windows[i][:80]}...")
                    print(f"    Окно {j}: {windows[j][:80]}...")
        else:
            print("Похожих пар не найдено")
    
    return all_results
```

```python
def analyze_all_results(all_results):
    
    successful_results = [r for r in all_results if r['similar_pairs_count'] > 0]
    
    if not successful_results:
        print("не найдено похожих пар")
        return
    print(f"Успешных экспериментов: {len(successful_results)} из {len(all_results)}")
    
    print("\nЛучгий результат (макс. jaccard):")
    successful_results.sort(key=lambda x: x['max_jaccard'], reverse=True)
    for i, result in enumerate(successful_results[:5]):
        p = result['params']
        print(f"{i+1}. J={result['max_jaccard']:.4f}")
        print(f"   Окна: {p['window_size']}, k={p['k']}, hashes={p['num_hashes']}, bands={p['bands']}")
        print(f"   Всего пар: {result['similar_pairs_count']}")
    
    print("\n лучший по количеству пар:")
    successful_results.sort(key=lambda x: x['similar_pairs_count'], reverse=True)
    for i, result in enumerate(successful_results[:3]):
        p = result['params']
        print(f"{i+1}. Пар: {result['similar_pairs_count']}")
        print(f"   Окна: {p['window_size']}, k={p['k']}, Макс.J: {result['max_jaccard']:.4f}")
    
    best_result = successful_results[0]
    print(f"\nлучший результат:")
    print(f"   Параметры: {best_result['params']}")
    print(f"   Макс. Jaccard: {best_result['max_jaccard']:.4f}")
    print(f"   Всего пар: {best_result['similar_pairs_count']}")
    
    if best_result['similar_pairs']:
        print(f"\n пары:")
        for i, j, jac in best_result['similar_pairs'][:3]:
            print(f"   Окна {i}-{j} | J={jac:.4f}")
```

```python
def lsh(signatures, bands):
    """Разрезает сигнатуры на куски (bands), и группирует индексы сигнатур по совпадению кусков"""
    buckets = defaultdict(list)
    band_length = len(signatures[0]) // bands
    
    for idx, sig in tqdm(enumerate(signatures)):
        for b in range(0, bands, band_length):
            start = b
            end = start + band_length
            band = tuple(sig[start:end])
            buckets[band].append(idx)
    return buckets
```
